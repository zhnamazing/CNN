```python
import numpy as np
import pandas as pd
import os

#中位数为26
def trans(str):
    num = int(str, 2)

    if num > 32767:
        # 负数，计算补码
        value =num-65536
    else:
        value = num
    return value

def trans40(str):
    num = int(str, 2)

    if num > 549755813887:
        # 负数，计算补码
        value =num-1099511627776
    else:
        value = num
    return value

def binary_to_decimal(number):

    if number > 32767:
        # 负数，计算补码
        value =number-65536
    else:
        value = number
    return value

def cut_sobel(dacc, bit):
    mask = (1 << (bit-15)) - 1  # 0b11111111111111111111111111

    temp_bits = (dacc >> 15) & mask  # 获取27-40位
    dacc_40 = (dacc >> (bit-1)) & 1  # 获取最高位
    dacc_li = dacc & ((1 << 15) - 1)  # 获取0-14位

    if temp_bits == 0 or temp_bits == (1 << (bit-15)) - 1:
        dcut = (dacc_40 << 15) | dacc_li
    else:
        dcut = 0b1000000000000000 if dacc_40 else 0b0111111111111111
    return (binary_to_decimal(dcut))

test_data_convert = list(map(trans, test_data))
test_data_np = np.array(test_data_convert)
IFM = test_data_np.reshape(14,14,8)##IFM是原输入图，按优先级reshape

mapp = np.zeros((8,14,14), dtype=np.int16)#8通道，14行14列
mapp = np.transpose(IFM, (2, 0, 1))#调整轴的顺序为通道、行、列

count = 0
for i in range(14):
    for j in range(14):
        for k in range(8):
            if mapp[k][i][j] == 0:
                count = count + 1
            
print('原输入特征图：')
print(mapp)

print('稀疏度：')
print(count/(14*14*8))

data = []
for channel in range(0,8):  # 遍历每个通道
    channel_data = mapp[channel]  # 获取当前通道的数据
    channel_df = pd.DataFrame(channel_data)  # 转换为 DataFrame
    channel_df['Channel'] = f'Channel {channel + 1}'  # 添加通道标识列
    data.append(channel_df)

# 将所有通道的数据合并成一个大的 DataFrame
combined_df = pd.concat(data, ignore_index=True)

# 将 DataFrame 写入 Excel 文件
combined_df.to_excel('C:\\Users\\zuohaonan\\Desktop\\PE2\\NN_accelerator\\output.xlsx', index=False)

kernel_convert = list(map(trans, kernel))
ker = np.zeros((32,200), dtype=np.int16)

Core = [np.zeros((8,5,5), dtype=int) for _ in range(32)]
kernel_re = [np.zeros((5,5,8), dtype=int) for _ in range(32)]

bias_convert = list(map(trans40, bias))
print('偏置:')
print(bias_convert)

for i in range(0,32):
    for j in range(0,200):
        ker[i][j] = kernel_convert[j + 200 * i]
        
for k in range(0,32):
    kernel_re[k] = ker[k].reshape(5,5,8)#行、列、通道
    Core[k] = np.transpose(kernel_re[k], (2, 0, 1))#调整轴的顺序为通道、行、列
    print(f"这是第{k}个卷积核:")
    print(Core[k])

post_MAC = [np.zeros((10, 10), dtype=int) for _ in range(32)]
mem = [np.zeros((10, 10), dtype=int) for _ in range(32)]
OFM = [np.zeros((5, 5), dtype=int) for _ in range(32)]

def conv(row_loc,col_loc,num):
    result = 0
    for i in range(0,5):
        for j in range(0,5):
            for k in range(0,8):
                result += (Core[num])[k][i][j]*mapp[k][row_loc+i][col_loc+j]

    mem[num][row_loc][col_loc] = result
    result += int(bias_convert[num])

    if result < 0:
        result = 0
            
    return result

def cut(dacc,bit):

    mask= (1 << (26-bit)) - 1  # 0b11111111111111111111111111
    
    temp_bits = (dacc>> (bit+15)) & mask#获取27-40位
    dacc_40 = (dacc >> 40) & 1#获取最高位
    dacc_li = (dacc>>bit) & ((1 << 15) - 1)#获取0-14位

    if temp_bits == 0 or temp_bits == (1 << (26-bit)) - 1:
        dcut = (dacc_40 << 15) | dacc_li
    else:
        dcut = 0b1000000000000000 if dacc_40 else 0b0111111111111111
    return (binary_to_decimal(dcut))

def max(a,b,c,d):
    if(a>=b):
        t1 = a
    else:
        t1 = b

    if(c>=d):
        t2 = c
    else:
        t2 = d

    if(t1>=t2):
        return t1
    else:
        return t2
        
def pool(row_loc,col_loc,num):
      
    result = max((post_MAC[num])[2*row_loc][2*col_loc],(post_MAC[num])[2*row_loc+1][2*col_loc],(post_MAC[num])[2*row_loc][2*col_loc+1],(post_MAC[num])[2*row_loc+1][2*col_loc+1])
    return result

for i in range(0,10):
    for j in range(0,10):
        for k in range(0,32):
            temp = conv(i,j,k)
            post_MAC[k][i][j] = cut(int(temp),0)

for k in range(0,32):
    print(f'这是第{k}张中间结果')
    print(mem[k])
    print('\n')

for i in range(0,5):
    for j in range(0,5):
        for k in range(0,32):
            OFM[k][i][j] = pool(i,j,k)

all_values = np.concatenate([arr.flatten() for arr in OFM])
non_zero_values = all_values[all_values != 0]  # 过滤非零值

# 计算统计值（包括全体的最大值和最小值）
max_value = np.max(all_values)  # 包含零的最大值
min_value = np.min(all_values)  # 包含零的最小值

# 计算非零值的平均值和中位数
if len(non_zero_values) > 0:
    mean_non_zero = np.mean(non_zero_values)
    median_non_zero = np.median(non_zero_values)
else:
    mean_non_zero = 0  # 如果无非零值，默认输出0
    median_non_zero = 0

# 输出结果
print(f"最大值（含零）: {max_value}")
print(f"最小值（含零）: {min_value}")
print(f"非零值的平均值: {mean_non_zero}")
print(f"非零值的中位数: {median_non_zero}")
limit = round(mean_non_zero/16)

with open('C:\\Users\\zuohaonan\\Desktop\\PE2\\NN_accelerator\\ofm_data02.txt', 'w') as file:
    for k in range(0,32):
        print(f'这是第{k}张输出图')
        print(OFM[k])
        print('\n')
        for i in range(0,5):
            for j in range(0,5):
                if OFM[k][i][j] < 0:
                    value = OFM[k][i][j] + 65536
                else:
                    value = OFM[k][i][j]
                        
                hex_value = format(value, '04X')
                file.write(hex_value + '\n')

with h5py.File('C:\\Users\\zuohaonan\\Desktop\\PE2\\NN_accelerator\\resultl2t3.h5', 'w') as hf:
    # 初始化result1数据集（1568个UTF-8编码的16位字符串）
    result1 = hf.create_dataset("result", 
                               shape=(800,), 
                               dtype=h5py.string_dtype(encoding='utf-8', length=16))  # UTF-8固定长度16字节
    
    index = 0  # 写入位置索引
    
    # 遍历OFM并填充result1
    
    for i in range(5):
        for j in range(5):
            for k in range(32):
                # 处理负数（转换为补码形式的正数）
                value = OFM[k][i][j]
                if value > (limit + 1):
                    value = int(value/limit)  # 16位补码转换
                
                # 转换为16位二进制字符串（UTF-8格式）
                binary_str = format(value, '016b')  # 如 "0000000000000000"
                result1[index] = binary_str
                index += 1

print("已写入完毕")
```
